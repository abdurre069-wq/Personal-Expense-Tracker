```python
import os
from datetime import datetime
```


### File Name Variable

- `FILE_NAME` is a variable that stores the name of the JSON file.
- `"expenses.json"` is the file where expense records will be stored.
- This file is used to save expense data permanently.
- The program can retrieve saved expenses when it runs again.


```python
# File to store expenses
FILE_NAME = "expenses.json"
```


### Load Expenses Function

- `def load_expenses():` defines a function to load saved expense records.
- `os.path.exists(FILE_NAME)` checks whether the expense file exists.
- `with open(FILE_NAME, "r")` opens the file in read mode.
- `json.load(file)` reads the JSON data and converts it into Python data.
- `return json.load(file)` returns the saved expense records.
- `return []` returns an empty list if the file does not exist.

**Purpose:**

This function allows the program to retrieve previously saved expenses whenever it starts.


```python
# Load expenses from file
def load_expenses():
    if os.path.exists(FILE_NAME):
        with open(FILE_NAME, "r") as file:
            return json.load(file)
    return []
```


### Save Expenses Function

- `def save_expenses(expenses):` defines a function to save expense records.
- `expenses` is the parameter that contains expense data.
- `with open(FILE_NAME, "w")` opens the JSON file in write mode.
- `"w"` means write mode. It creates a new file if it does not exist.
- `json.dump(expenses, file, indent=4)` saves expense data in JSON format.
- `indent=4` makes the JSON data properly formatted and easier to read.

**Purpose:**

This function saves expense records permanently so they can be accessed when the program runs again.


```python
# Save expenses to file
def save_expenses(expenses):
    with open(FILE_NAME, "w") as file:
        json.dump(expenses, file, indent=4)
```


### Add Expense Function

- `def add_expense(expenses):` defines a function to add a new expense.
- `print()` displays the Add New Expense heading.
- `input()` asks the user to enter the expense name.
- `try` handles possible input errors.
- `float()` converts the expense amount into a decimal number.
- `if amount <= 0` checks whether the entered amount is valid.
- `return` exits the function if the amount is invalid.
- `except ValueError` handles errors when the user enters non-numeric input.
- `expense` is a dictionary that stores the expense name, amount, and date.
- `datetime.now()` gets the current date and time.
- `strftime("%Y-%m-%d")` formats the date.
- `expenses.append(expense)` adds the new expense to the list.
- `save_expenses(expenses)` saves the updated expenses into the JSON file.
- `print()` displays a success message.

**Purpose:**

This function allows users to enter new expenses, validates the input, records the current date, and saves the expense details permanently.


```python
 #Add a new expense
def add_expense(expenses):
    print("\n--- Add New Expense ---")

    name = input("Enter expense name: ")

    try:
        amount = float(input("Enter expense amount (PKR): "))

        if amount <= 0:
            print("Amount must be greater than zero!")
            return

    except ValueError:
        print("Please enter a valid number!")
        return

    expense = {
        "name": name,
        "amount": amount,
        "date": datetime.now().strftime("%Y-%m-%d")
    }

    expenses.append(expense)
    save_expenses(expenses)

    print("Expense added successfully!")
```


### View Expenses Function

- `def view_expenses(expenses):` defines a function to display all recorded expenses.
- `print()` displays the All Expenses heading.
- `if not expenses` checks whether the expenses list is empty.
- `print()` displays a message if no expenses are recorded.
- `return` exits the function when the list is empty.
- `for` loop iterates through all recorded expenses.
- `enumerate(expenses, start=1)` provides an index for each expense, starting from 1.
- `expense` represents an individual expense dictionary.
- `f-string` formats and displays expense information.
- `expense['name']` retrieves the expense name.
- `expense['amount']` retrieves the expense amount.
- `:.2f` formats the amount to two decimal places.
- `expense['date']` retrieves the expense date.

**Purpose:**

This function displays all recorded expenses with their names, amounts, and dates. It also handles empty expense lists.


```python
# View all expenses
def view_expenses(expenses):
    print("\n--- All Expenses ---")

    if not expenses:
        print("No expenses recorded yet.")
        return

    for index, expense in enumerate(expenses, start=1):
        print(
            f"{index}. {expense['name']} - "
            f"PKR {expense['amount']:.2f} - "
            f"{expense['date']}"
        )
```


### Total Expenses Function

- `def total_expenses(expenses):` defines a function to calculate total spending.
- `sum()` calculates the sum of all expense amounts.
- `expense["amount"]` retrieves the amount from each expense dictionary.
- `for expense in expenses` iterates through all recorded expenses.
- `total` stores the calculated sum of all expenses.
- `print()` displays the Total Expenses heading.
- `f-string` formats and displays the total spending.
- `:.2f` formats the total amount to two decimal places.

**Purpose:**

This function calculates and displays the total amount of money spent on all recorded expenses.


```python
# Calculate total expenses
def total_expenses(expenses):
    total = sum(expense["amount"] for expense in expenses)

    print("\n--- Total Expenses ---")
    print(f"Total Spending: PKR {total:.2f}")
```


### Main Function

- `def main():` defines the main function that controls the entire program.
- `load_expenses()` loads previously saved expense records.
- `expenses` stores the loaded expense data.
- `while True` keeps the program running continuously until the user exits.
- `print()` displays the main menu and available options.
- `input()` asks the user to select an option.
- `if` checks whether the user selected option 1.
- `elif` checks the other available menu options.
- `add_expense(expenses)` calls the function to add a new expense.
- `view_expenses(expenses)` calls the function to display all expenses.
- `total_expenses(expenses)` calls the function to calculate total spending.
- `break` terminates the loop and exits the program.
- `else` handles invalid menu choices.

**Purpose:**

The main function controls the program, displays the menu, processes user choices, and calls the appropriate functions.

### Program Entry Point

- `if __name__ == "__main__":` checks whether the Python file is being executed directly.
- `main()` starts the program by calling the main function.

**Purpose:**

This condition ensures that the main program runs when the file is executed directly.


```python

# Main program
def main():
    expenses = load_expenses()

    while True:
        print("\n________________")
        print("     PERSONAL EXPENSE TRACKER")
        print("===============")
        print("1. Add Expense")
        print("2. View Expenses")
        print("3. Total Expenses")
        print("4. Exit")
        print("============")

        choice = input("Enter your choice (1-4): ")

        if choice == "1":
            add_expense(expenses)

        elif choice == "2":
            view_expenses(expenses)

        elif choice == "3":
            total_expenses(expenses)

        elif choice == "4":
            print("Thank you for using Expense Tracker!")
            break

        else:
            print("Invalid choice! Please select 1-4.")


# Start the program
if __name__ == "__main__":
    main()
```
